1. <h3 style="color: plum;"><b>Write a Python function to find the maximum and minimum numbers from a sequence of numbers.  (Note: Do not use built-in functions.)

```python
def find_max_min(numbers):
    if not numbers:
        return None, None
    max_num = numbers[0]
    min_num = numbers[0]
    for num in numbers:
        if num > max_num:
            max_num = num
        elif num < min_num:
            min_num = num
    return max_num, min_num
numbers = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]
max_number, min_number = find_max_min(numbers)
print(f"Maximum number: {max_number}")
print(f"Minimum number: {min_number}")
```

2. <h3 style="color: plum;"><b>Write a Python function that takes a positive integer and returns the sum of the cube of all the positive integers smaller than the specified number.

```python
def sum_of_cubes(n):
    if n <= 0:
        return 0
    total_sum = 0
    for i in range(1, n):
        total_sum += i ** 3
    return total_sum
number = 5
result = sum_of_cubes(number)
print(f"The sum of the cubes of all positive integers smaller than {number} is: {result}")
```

3. <h3 style="color: plum;"><b>Write a Python function to print 1 to n using recursion. (Note: Do not use loop)

```python
def print_numbers(n):
    if n <= 0:
        return  
    print_numbers(n - 1)  
    print(n)
n = 5
print(f"Numbers from 1 to {n}:")
print_numbers(n)
```

4. <h3 style="color: plum;"><b>Write a recursive function to print Fibonacci series upto n terms

```python
def fibonacci(n, a=0, b=1):
    if n <= 0:
        return
    print(a, end=' ')
    fibonacci(n - 1, b, a + b)
n_terms = 10
print(f"Fibonacci series up to {n_terms} terms:")
fibonacci(n_terms)
```

5. <h3 style="color: plum;"><b>Write a lambda function to find volume of cone. 

```python
volume_of_cone = lambda r, h: (1/3) * 3.14159 * r**2 * h
radius = 5
height = 10
volume = volume_of_cone(radius, height)
print(f"The volume of the cone with radius {radius} and height {height} is: {volume}")
```

6. <h3 style="color: plum;"><b>Write a lambda function which gives tuple of max and min from a list. 
Sample input: [10, 6, 8, 90, 12, 56] 

Sample output: (90,6) 

```python
max_min_tuple = lambda lst: (max(lst), min(lst)) if lst else (None, None)
numbers = [10, 6, 8, 90, 12, 56]
result = max_min_tuple(numbers)
print(f"The maximum and minimum numbers from the list {numbers} are: {result}")
```

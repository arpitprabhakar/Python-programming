6. Write a program to compute the length of the hypotenuse (c) of a right triangle
using Pythagoras theorem.


```python
a= float(input("Enter base:"))
b=float(input("Enter height:"))
c=(a*a+b*b)**0.5
print("hypo: ",c)
```

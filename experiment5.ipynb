1. <h3 style="color: plum;"><b>Write a program to count and display the number of capital letters in a given string.

```python
string = input("Enter a string: ")
capital_count = sum(1 for char in string if char.isupper())
print(f"The number of capital letters in the string is: {capital_count}")
```

2. <h3 style="color: plum;"><b>Count total number of vowels in a given string

```python
string = input("Enter a string: ")
num = 0
for word in string:
    if word in 'aeiouAEIOU':
        num += 1
print(f"Total number of vowels: {num}")
```

4. <h3 style="color: plum;"><b>WAP to enter a string and a substring. You have to print the number of times that the substring occurs in the given string. String traversal will take place from left to right, not from right to left.

5. <h3 style="color: plum;"><b>	Given a string containing both upper and lower-case alphabets. Write a Python program to count the number of occurrences of each alphabet (case insensitive) and display the same. 
Sample Input 

ABaBCbGc 

Sample Output 

2A 

3B 

2C 

1G 

```python
string = input("Enter a string: ")
string = string.upper()
alphabet_count = {}
for char in string:
    if char.isalpha():
        if char in alphabet_count:
            alphabet_count[char] += 1
        else:
            alphabet_count[char] = 1
for char, count in alphabet_count.items():
    print(f"{count}{char}")
```

6. <h3 style="color: plum;"><b>Program to count the number of unique words in a given sentence using sets

```python
sentence = input("Enter a sentence: ")
words = sentence.split()
unique_words = set(words)
print(f"The number of unique words in the sentence is: {len(unique_words)}")
```

7. <h3 style="color: plum;"><b>Create 2 sets s1 and s2 of n fruits each by taking input from user and find: 

a. intersection of s1 and s2

b. difference in s1

c.	Union of s1 and s2

```python
n = int(input("Enter number of fruits in each set: "))
print(f"Enter {n} fruits for Set 1:")
s1 = set()
for _ in range(n):
    fruit = input("  Fruit: ").strip().lower()
    s1.add(fruit)
print(f"Enter {n} fruits for Set 2:")
s2 = set()
for _ in range(n):
    fruit = input("  Fruit: ").strip().lower()
    s2.add(fruit)
print(f"\nSet 1: {s1}")
print(f"Set 2: {s2}")
print(f"\na. Fruits in both s1 and s2: {s1 & s2}")
print(f"b. Fruits only in s1 but not in s2: {s1 - s2}")
all_fruits = s1 | s2
print(f"c. Count of all fruits from s1 and s2 (union): {len(all_fruits)}")
print(f"   All fruits: {all_fruits}")
```

8. <h3 style="color: plum;"><b>Take two sets and apply various set operations on them: 
S1 = {Red, yellow, orange, blue} 
S2 = {violet, blue, purple}

```python
S1 = {"Red", "yellow", "orange", "blue"}
S2 = {"violet", "blue", "purple"}
print(f"Union of S1 and S2: {S1.union(S2)}")
print(f"Intersection of S1 and S2: {S1.intersection(S2)}")
print(f"Difference of S1 and S2 (S1 - S2): {S1.difference(S2)}")
print(f"Difference of S1 and S2 (S2 - S1): {S2.difference(S1)}")
print(f"Symmetric Difference of S1 and S2: {S1.symmetric_difference(S2)}")
```

```python
# Mapping: digit → letter (0=a, 1=b, ..., 9=j)
# Encrypt: each digit of the ID becomes a letter
# Decrypt: each letter becomes a digit back
# Add:     combine two encrypted strings into one

def encrypt(id_number):
    result = ""
    for digit in str(id_number):
        result += chr(ord('a') + int(digit))  # 0→a, 1→b, ..., 9→j
    return result

def decrypt(encrypted):
    result = ""
    for letter in encrypted:
        result += str(ord(letter) - ord('a'))  # a→0, b→1, ..., j→9
    return int(result)

def add(enc1, enc2):
    return enc1 + enc2  # concatenate two encrypted strings

# ── Demo ──────────────────────────────────────────────────────────────────────

id1 = 5029
id2 = 1234

# Step 1: Encrypt
enc1 = encrypt(id1)
enc2 = encrypt(id2)
print(f"Encrypt {id1} → '{enc1}'")
print(f"Encrypt {id2} → '{enc2}'")

# Step 2: Decrypt
print(f"\nDecrypt '{enc1}' → {decrypt(enc1)}")
print(f"Decrypt '{enc2}' → {decrypt(enc2)}")

# Step 3: Add (operator overloading via + on strings)
combined = add(enc1, enc2)
print(f"\nAdd '{enc1}' + '{enc2}' → '{combined}'")
print(f"Decrypt combined '{combined}' → {decrypt(combined)}")
```

```python
with open("table.txt", "w") as f:
    f.write("Name, Address\n")
    f.write("Arpit, Lucknow\n")
    f.write("sanyam, bihar\n")
    f.write("akshat, kanpur\n")
    f.write("avadhya, bangalore\n")


data_store = {}
with open("table.txt", "r") as f:
    lines = f.read().splitlines()

for i in lines[1:]:
    name, city = i.split(",")
    name = name.strip()
    city = city.strip()
    
    if city not in data_store:
        data_store[city] = {"count": 0, "students": []}
    
    data_store[city]["count"] += 1
    data_store[city]["students"].append(name)

print(f"{'City':<15} | {'Count':<8} | {'Students'}")
print("-" * 45)

for city, info in data_store.items():
    students_str = ", ".join(info["students"])
    print(f"{city:<15} | {info['count']:<8} | {students_str}")
```

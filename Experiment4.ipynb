1. <h3 style="color: plum;"><b>Find a factorial of given number.

```python

num = int(input("Enter a number: "))
factorial = 1
if num < 0:
    print("Sorry, factorial does not exist for negative numbers")
elif num == 0:
    print("The factorial of 0 is 1")
else:
    for i in range(1, num + 1):
        factorial *= i
    print("The factorial of", num, "is", factorial)
    
```

2. <h3 style="color: plum;"><b> Find whether the given number is Armstrong number

```python


num = int(input("Enter a number: "))
order = len(str(num))
sum = 0
temp = num
while temp > 0:
    digit = temp % 10
    sum += digit ** order
    temp //= 10
if num == sum:
    print(num, "is an Armstrong number")
else:
    print(num, "is not an Armstrong number")
    
```

3. <h3 style="color: plum;"><b>Print Fibonacci series up to given term

```python
n = int(input("Enter the number of terms: "))
a, b = 0, 1
count = 0
if n <= 0:
    print("Please enter a positive integer")
elif n == 1:
    print("Fibonacci sequence up to", n, ":")
    print(a)
else:
    print("Fibonacci sequence up to", n, ":")
    while count < n:
        print(a)
        a, b = b, a + b
        count += 1
        
```

4. <h3 style="color: plum;"><b> Write a program to find if given number is prime number or not. 

```python

num = int(input("Enter a number: "))
if num > 1:
    for i in range(2, int(num/2) + 1):
        if (num % i) == 0:
            print(num, "is not a prime number")
            break
    else:
        print(num, "is a prime number")
```

5. <h3 style="color: plum;"><b> Check whether given number is palindrome or not

```python
num = int(input("Enter a number: "))
original = num
reverse = 0
while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num //= 10
 
if original == reverse:
    print(f"{original} is a Palindrome number.")
else:
    print(f"{original} is NOT a Palindrome number.")
```

6. <h3 style="color: plum;"><b>Write a program to print sum of digits


```python

num = int(input("Enter a number: "))
sum = 0
while num > 0:
    digit = num % 10
    sum += digit
    num //= 10
print("The sum of the digits is:", sum)
```

7. <h3 style="color: plum;"><b>Count and print all numbers divisible by 5 or 7 between 1 to 100. 


```python

count = 0
for i in range(1, 101):
    if i % 5 == 0 or i % 7 == 0:
        print(i)
        count += 1
print("Total numbers divisible by 5 or 7 between 1 and 100:", count)
```

8. <h3 style="color: plum;"><b> Convert all lower cases to upper case in a string


```python

string = input("Enter a string: ")
upper_string = ""
for char in string:
    if char.islower():
        upper_string += char.upper()
    else:
        upper_string += char
print("String with lowercase converted to uppercase:", upper_string)
```
